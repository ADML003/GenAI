```python
import os
from dotenv import load_dotenv
load_dotenv()
```

### HuggingFaceEndpoint
#### How to Access HuggingFace Models with API
There are also two ways to use this class. You can specify the model with the repo_id parameter. Those endpoints use the serverless API, which is particularly beneficial to people using pro accounts or enterprise hub. Still, regular users can already have access to a fair amount of request by connecting with their HF token in the environment where they are executing the code.

```python
from langchain_huggingface import HuggingFaceEndpoint, HuggingFacePipeline
```

```python
# Using Google's Flan-T5 which is reliable for text generation
repo_id="google/flan-t5-large"
llm=HuggingFaceEndpoint(
    repo_id=repo_id,
    temperature=0.7,
    max_new_tokens=512,  # Use max_new_tokens instead of max_length
    huggingfacehub_api_token=os.getenv("HF_TOKEN")
)
llm
```

```python
# Note: This may fail with API issues, use local pipeline instead
# llm.invoke("What is generative ai?")
```

```python
# Alternative: Using HuggingFacePipeline for local execution
# This is more reliable and doesn't require API access

# Create a local pipeline (this will download the model locally)
local_llm = HuggingFacePipeline.from_model_id(
    model_id="google/flan-t5-small",  # Using smaller model for faster loading
    task="text2text-generation",
    model_kwargs={"temperature": 0.7, "max_length": 100}
)

print("Local LLM created successfully!")
local_llm
```

```python
# Test the local LLM
response = local_llm.invoke("What is generative AI?")
print("Response:", response)
```

## Issues with HuggingFaceEndpoint and Solutions

### Problem
The original `HuggingFaceEndpoint` was failing with a `StopIteration` error because:

1. **Model Availability**: Some models may not be available through the Hugging Face Inference API
2. **Provider Mapping**: The API couldn't determine the appropriate provider for the model
3. **API Limitations**: Free tier limitations or model restrictions

### Solutions

#### Option 1: HuggingFacePipeline (Local Execution) ✅ **Working**
- Downloads and runs models locally
- No API limits
- Better for development and testing
- Uses your local compute resources

#### Option 2: Different Models with HuggingFaceEndpoint
- Try models that are confirmed to work with the Inference API
- Use smaller, more common models
- Check model documentation for API support

#### Option 3: Use Different Parameters
- Avoid conflicting parameters like `max_length` vs `max_new_tokens`
- Use simpler configurations for testing

```python
# Example with embeddings
from langchain_community.embeddings import HuggingFaceBgeEmbeddings

model_name = "BAAI/bge-small-en"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": True}
hf = HuggingFaceBgeEmbeddings(
    model_name=model_name, model_kwargs=model_kwargs, encode_kwargs=encode_kwargs
)
```

```python
embedding = hf.embed_query("hi this is harrison")
print(f"Embedding dimension: {len(embedding)}")
print(f"First 5 values: {embedding[:5]}")
```
