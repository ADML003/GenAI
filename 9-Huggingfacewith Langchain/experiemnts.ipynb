## Issues with HuggingFaceEndpoint and Solutions

### Problem
The original `HuggingFaceEndpoint` was failing with a `StopIteration` error because:

1. **Model Availability**: Some models may not be available through the Hugging Face Inference API
2. **Provider Mapping**: The API couldn't determine the appropriate provider for the model
3. **API Limitations**: Free tier limitations or model restrictions

### Solutions

#### Option 1: HuggingFacePipeline (Local Execution) ✅ **Working**
- Downloads and runs models locally
- No API limits
- Better for development and testing
- Uses your local compute resources

#### Option 2: Different Models with HuggingFaceEndpoint
- Try models that are confirmed to work with the Inference API
- Use smaller, more common models
- Check model documentation for API support

#### Option 3: Use Different Parameters
- Avoid conflicting parameters like `max_length` vs `max_new_tokens`
- Use simpler configurations for testing

## New Functionality 1: Document Q&A with Vector Store

Let's build a document Q&A system using the embeddings we created. This will allow us to ask questions about documents.

```python
# Install required packages for vector stores
!pip install faiss-cpu langchain-community
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain.chains import RetrievalQA

# Sample documents
documents = [
    "Generative AI is a type of artificial intelligence that can create new content like text, images, and music.",
    "LangChain is a framework for developing applications powered by language models.",
    "Hugging Face provides thousands of pre-trained models for natural language processing tasks.",
    "Machine learning is a subset of artificial intelligence that enables systems to learn from data.",
    "Deep learning uses neural networks with multiple layers to learn complex patterns in data.",
    "Natural language processing helps computers understand and generate human language.",
    "Transformers are a type of neural network architecture that revolutionized NLP.",
    "BERT and GPT are popular transformer-based models for various NLP tasks."
]

# Split documents into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
texts = text_splitter.create_documents(documents)

print(f"Created {len(texts)} document chunks")
print(f"Sample chunk: {texts[0].page_content}")
```

```python
# Create vector store using the embeddings we created earlier
vectorstore = FAISS.from_documents(texts, hf)
print("Vector store created successfully!")

# Create a retriever
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

# Test retrieval
query = "What is generative AI?"
docs = retriever.get_relevant_documents(query)
print(f"\nTop 3 relevant documents for: '{query}'")
for i, doc in enumerate(docs, 1):
    print(f"\n{i}. {doc.page_content}")
```

```python
# Create a Q&A chain with the vector store and LLM
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)

# Ask questions
question = "What is LangChain used for?"
result = qa_chain({"query": question})

print(f"Question: {question}")
print(f"\nAnswer: {result['result']}")
print(f"\nSource Documents:")
for i, doc in enumerate(result['source_documents'], 1):
    print(f"{i}. {doc.page_content}")
```

## New Functionality 2: Conversational Chain with Memory

Add memory to your conversations so the model remembers previous interactions.

```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory

# Create a conversation chain with memory
memory = ConversationBufferMemory()
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True
)

print("Conversational chain created with memory!")
```

```python
# Have a conversation
response1 = conversation.predict(input="My name is Alex and I love machine learning")
print(f"Response 1: {response1}\n")

response2 = conversation.predict(input="What is my name?")
print(f"Response 2: {response2}\n")

response3 = conversation.predict(input="What do I love?")
print(f"Response 3: {response3}")
```

## New Functionality 3: Text Summarization

Use the models to summarize long texts efficiently.

```python
from langchain.chains.summarize import load_summarize_chain
from langchain.docstore.document import Document

# Sample long text
long_text = """
Artificial Intelligence (AI) has revolutionized the way we interact with technology. 
Machine learning, a subset of AI, enables computers to learn from data without explicit programming.
Deep learning, which uses neural networks with multiple layers, has achieved remarkable results in 
image recognition, natural language processing, and speech recognition.

Large Language Models (LLMs) like GPT and BERT have transformed natural language processing tasks.
These models are pre-trained on massive amounts of text data and can be fine-tuned for specific tasks.
The transformer architecture, introduced in 2017, is the foundation for most modern LLMs.

Generative AI has opened new possibilities for content creation, from writing articles to generating 
images and music. Tools like ChatGPT, DALL-E, and Midjourney have made AI accessible to the general public.
However, ethical considerations around AI safety, bias, and misinformation remain important challenges 
that need to be addressed as the technology continues to evolve.
"""

# Create a document
doc = Document(page_content=long_text)

# Load summarization chain
summarize_chain = load_summarize_chain(llm, chain_type="stuff")

# Generate summary
summary = summarize_chain.run([doc])
print("Original text length:", len(long_text))
print(f"\nSummary:\n{summary}")
```

## New Functionality 4: Multi-Model Comparison

Compare responses from different Hugging Face models to see which performs best for your task.

```python
import time

# Define models to compare
models_to_compare = [
    "google/flan-t5-large",
    "google/flan-t5-base",
    "mistralai/Mistral-7B-Instruct-v0.3"
]

# Test prompt
test_prompt = "Explain what machine learning is in simple terms."

results = {}

for model_id in models_to_compare:
    try:
        print(f"\nTesting model: {model_id}")
        
        # Create model instance
        test_llm = HuggingFaceEndpoint(
            repo_id=model_id,
            temperature=0.7,
            max_new_tokens=150,
            huggingfacehub_api_token=os.getenv("HF_TOKEN")
        )
        
        # Time the response
        start_time = time.time()
        response = test_llm.invoke(test_prompt)
        end_time = time.time()
        
        results[model_id] = {
            "response": response,
            "time": round(end_time - start_time, 2)
        }
        
        print(f"✓ Success - Response time: {results[model_id]['time']}s")
        
    except Exception as e:
        results[model_id] = {
            "response": f"Error: {str(e)}",
            "time": 0
        }
        print(f"✗ Failed: {str(e)}")
```

```python
# Display comparison results
print("\n" + "="*80)
print("MODEL COMPARISON RESULTS")
print("="*80)
print(f"\nPrompt: {test_prompt}\n")

for model_id, result in results.items():
    print(f"\n{'─'*80}")
    print(f"Model: {model_id}")
    print(f"Response Time: {result['time']}s")
    print(f"Response:\n{result['response']}")
    print(f"{'─'*80}")
```

## New Functionality 5: Advanced Prompt Engineering with Few-Shot Learning

Improve model responses using few-shot examples and structured prompts.

```python
from langchain.prompts import FewShotPromptTemplate, PromptTemplate

# Define examples for few-shot learning
examples = [
    {
        "question": "What is Python?",
        "answer": "Python is a high-level, interpreted programming language known for its simplicity and readability."
    },
    {
        "question": "What is JavaScript?",
        "answer": "JavaScript is a versatile programming language primarily used for web development and creating interactive websites."
    },
    {
        "question": "What is SQL?",
        "answer": "SQL (Structured Query Language) is a language designed for managing and querying relational databases."
    }
]

# Create an example template
example_template = """
Question: {question}
Answer: {answer}
"""

example_prompt = PromptTemplate(
    input_variables=["question", "answer"],
    template=example_template
)

# Create the few-shot prompt template
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    prefix="Answer the following question in a clear and concise manner, similar to the examples below:",
    suffix="Question: {input}\nAnswer:",
    input_variables=["input"]
)

print("Few-shot prompt template created!")
print("\nSample prompt:")
print(few_shot_prompt.format(input="What is Java?"))
```

```python
# Test the few-shot prompt with LLM
test_question = "What is Ruby?"
formatted_prompt = few_shot_prompt.format(input=test_question)

response = llm.invoke(formatted_prompt)
print(f"Question: {test_question}")
print(f"Answer: {response}")
```
