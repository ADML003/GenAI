```python
import os
from dotenv import load_dotenv
load_dotenv()
# --- IGNORE ---
```

```python
groq_api_key=os.getenv("GROQ_API_KEY")
```

```python
!pip install langchain_groq
```

```python
from langchain_groq import ChatGroq
model=ChatGroq(model="Gemma2-9b-It", groq_api_key=groq_api_key)
model
```

```python
!pip install langchain_core
```

```python
from langchain_core.messages import HumanMessage,SystemMessage

messages=[
    SystemMessage(content="translate the follwoing from english to french"),
    HumanMessage(content="hello, how are you?")
]

res=model.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
output=parser.invoke(res)
```

```python
output
```

```python
## using LCEL - chain the component
chain=model | parser
chain.invoke(messages)
```

```python
## Prompt templates

from langchain_core.prompts import   ChatPromptTemplate
```

```python
generic="translate the following from english to {language}"

prompt=ChatPromptTemplate.from_messages([
    ("system",generic),("user","{text}")
])

```

```python
result=prompt.invoke({"language":"french","text":"i love tesss"})
```

```python
result
```

```python
result.to_messages()
```

```python
chain=prompt | model | parser
chain.invoke({"language":"french","text":"Paris is. the city of Love and lights"})
```
