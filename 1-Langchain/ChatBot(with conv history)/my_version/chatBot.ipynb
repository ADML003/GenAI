```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory

from langchain_core.runnables.history import RunnableWithMessageHistory

store={} # dictonary to store the session history
```

```python
#function to get the session history
def get_session_history(session_id:str)-> BaseChatMessageHistory: #return type is BaseChatMessageHistory
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history=RunnableWithMessageHistory(model,get_session_history) 
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
res=with_message_history.invoke([HumanMessage(content="hi tesss here, i was developed by Darius Tanz of Tanz Industries")],config=config)
res
```

```python
res.content
```

```python
#changing the session id with config

config1={"configurable":{"session_id":"chat2"}}
res1=with_message_history.invoke([HumanMessage(content="whats my name?")],config=config1)
```

```python
res1 #differnet session, so it does not remember the previous context
```

```python
config1={"configurable":{"session_id":"chat2"}}
res1=with_message_history.invoke([HumanMessage(content="my name. is AxM")],config=config1)
```

```python
res1=with_message_history.invoke([HumanMessage(content=" what is my name ")],config=config1)

res1
```

## Prompt Templates

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant.Answer all the question to the nest of your ability"),
        MessagesPlaceholder(variable_name="messages")
    ]
)
```

```python
chain=prompt | model 
```

```python
chain.invoke({"messages":[HumanMessage(content="hi tesss here, i was developed by Darius Tanz of Tanz Industries")]})
```

```python
with_history=RunnableWithMessageHistory(chain,get_session_history)

config={"configurable":{"session_id":"chat3"}}

res2=with_history.invoke([HumanMessage(content="hi tesss here, i was developed by Darius Tanz of Tanz Industries")],config=config)
```

```python
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant.Answer all the question to the nest of your ability in {language}"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain=prompt | model

```

```python
res= chain.invoke({"messages":[HumanMessage(content="hi tesss here, i was developed by Darius Tanz of Tanz Industries")],"language":"Hindi"})

res.content
```

```python
from langchain_core.messages import SystemMessage,trim_messages
```

```python
trimmer=trim_messages(
    max_tokens=20,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
messages = [
    SystemMessage(content="you're a good assistant"),
    HumanMessage(content="hi! I'm bob"),
    AIMessage(content="hi!"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="nice"),
    HumanMessage(content="whats 2 + 2"),
    AIMessage(content="4"),
    HumanMessage(content="thanks"),
    AIMessage(content="no problem!"),
    HumanMessage(content="having fun?"),
    AIMessage(content="yes!"),
]
```

```python
trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough
```

```python
chain=(
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer)|prompt|model
)
res= chain.invoke({"messages":messages +[HumanMessage(content= "what math problem did i ask")],"language":"English"})

res.content
```

```python
with_history=RunnableWithMessageHistory(chain,get_session_history,input_messages_key="messages")
```

```python
config={"configurable":{"session_id":"chat6"}}
res=with_history.invoke({"messages":messages +[HumanMessage(content= "what math problem did i ask")],"language":"English"},config=config)
res.content
```
