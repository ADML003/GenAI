```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader = TextLoader('speech.txt')
docs=loader.load()

text_splitter=CharacterTextSplitter(chunk_size=200, chunk_overlap=20)

docss=text_splitter.split_documents(docs)
```

```python
docss
```

```python
embeddings=OllamaEmbeddings(model='llama3.2:latest')
db=FAISS.from_documents(docss,embeddings)
```

```python
query="What does Steve Jobs mean by 'connecting the dots looking backwards'?"

doc=db.similarity_search(query)
doc
```

```python
doc[0].page_content
```

```python
# as a retriever

retriever=db.as_retriever()
docs=retriever.invoke(query)
```

```python
docs[0].page_content
```

```python
docs_score=db.similarity_search_with_score(query)
docs_score
```

```python
#using embeddings 

query_embedding = embeddings.embed_query(query)
query_embedding
```

```python
db.similarity_search_by_vector(query_embedding)
```

```python
#. saving  and loading the vector db

db.save_local("faiss_index")
```

```python


new_db=FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)

docs=new_db.similarity_search(query)
docs
```
